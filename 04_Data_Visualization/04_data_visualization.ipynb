```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import ticker
import os
```

```python
src_files = os.path.join('data' , 'raw' , 'EPA_fuel_economy.csv')
image_dir = ('images')
```

```python
df = pd.read_csv(src_files)
df.head()
```

```python
df.info()
```

```python
plt.hist(df['comb08'])
```

```python
plt.hist(df['comb08']);
```

```python
plt.hist(df['comb08'])
plt.xlabel('Combianed fuel economy(Mile per Gallon)')
plt.ylabel('Numbers of Cares')
plt.title('Combianed fuel economy')
plt.show()
```

```python
fig , ax = plt.subplots()
ax.hist(df['comb08']);
```

```python
fig , ax = plt.subplots()
ax.hist(df['comb08'])
ax.set_xlabel('Combianed fuel economy(Mile per Gallon)')
ax.set_ylabel('Numbers of Cares')
ax.set_title('Combianed fuel economy')
plt.show()
```

```python
fig , ax = plt.subplots()
ax.hist(df['comb08'])
ax.set(
    xlabel = 'Combianed fuel economy(Mile per Gallon' ,
    ylabel = 'Numbers of Cares' ,
    title = 'Combianed fuel economy'
    
);
```

```python
fig , ax = plt.subplots()
ax.hist(df['comb08'] , range = (10 , 50));
```

```python
fig , ax = plt.subplots()
ax.hist(df['comb08'] , range = (10 , 50) , histtype = 'step' , orientation = 'horizontal');
```

```python
fig , ax = plt.subplots()
ax.hist(df['comb08'] , range = (10 , 50) , bins = 5 );
```

```python
fig , ax = plt.subplots()
ax.hist(df['comb08'] , range = (10 , 50) , bins = 50 );
```
