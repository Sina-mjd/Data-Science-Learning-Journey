```python
def quicksort(arr):
    if len(arr) <= 1:
        return arr
    arr = numpy.array(arr)
    p = arr [len(arr) // 2]
    l = [x for x in arr if x < p]
    m =  [x for x in arr if x == p]
    r = [x for x in arr if x > p]
    return quicksort(l) + m + quicksort(r)
```

```python
h = [5 , 10 , 1 , 13 , 9 , 4 , 3 , 0]
quicksort(h)
```
