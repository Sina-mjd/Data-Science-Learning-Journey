```python
from scipy.stats import poisson
from scipy.stats import norm
from matplotlib import pyplot as plt 
import pandas as pd
```

```python
p = [0.10 , 0.41 , 0 , 0.334]
q = [0.11 , 0.45 , 0.12 , 0.345]
```

```python
def kl(p , q) :
    result = np.sum(np.where(np.logical_and(p != 0 ,q != 0 ), p * np.log(p / q) , 0))
    return result
```

```python
x = np.arange(-10 , 10 , 0.001)
len(x)
```

```python
p = norm.pdf( x , 0 , 2)
q = norm.pdf(x , 2 , 3)
plt.plot(x , p , c = 'blue')
plt.plot(x , q , c = "red")
```

```python
kl(p , q)
```

```python
from scipy.stats import entropy
```

```python
def Js (p , q):
    p1 = p / np.linalg.norm(p , ord = 1)
    q1 = q / np.linalg.norm(q , ord = 1)
    m = 0.5 * (p1 + q1)
    r = 0.5 * (entropy(p1 , m) + entropy(q1 , m))
    return r
```

```python
whData = pd.read_csv('weight-height.csv')
whData.head()
```

```python
ideal = norm.rvs(size = len(whData['Height']) , loc = whData['Height'].mean() , scale = whData['Height'].std())
icount , idevision = np.histogram(ideal ,bins = 100)
```

```python
count , devision = np.histogram(whData['Height'] , bins = 100)
```

```python
Js(count ,icount)
```
